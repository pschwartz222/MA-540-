```python
import numpy as np
```

```python
Omega = [[i,j] for i in range(1,7) for j in range(1,7)]
```

```python
print(len(Omega))
```

```python
def X(roll):
  return(sum(roll))
```

```python
print(Omega[1])
print(X(Omega[1]))
```

```python
Image_X = []

for roll in Omega:
  x = X(roll)
  if x not in Image_X:
    Image_X.append(x)

print(Image_X)
```

```python
def X_inv(n):
  preimage_n = []
  for roll in Omega:
    if X(roll) == n:
      preimage_n.append(roll)
  return(preimage_n)
```

```python
for n in Image_X:
  E = X_inv(n)
  print(f"n = {n}, X^{-1}(n)= {E}")
  print(f"P(X=n) = {len(E)}/{len(Omega)}")
  print()
```
